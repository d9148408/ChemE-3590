## 1. 生成模擬數據：反應器操作數據

我們將模擬一個化工連續攪拌槽反應器 (CSTR) 的操作數據，包含：
- **正常操作數據**（95%）：反應器在穩定操作條件下的數據
- **異常操作數據**（5%）：包含多種異常情境
  - **溫度失控**：溫度急劇升高
  - **壓力異常**：壓力異常波動
  - **流量異常**：進料或冷卻流量異常
  - **攪拌故障**：攪拌速度異常
  - **產物濃度異常**：產物濃度異常偏離

**監控變數**：
- Temperature (°C)：反應器溫度
- Pressure (bar)：反應器壓力
- Feed_Flow (m³/h)：進料流量
- Cooling_Flow (m³/h)：冷卻水流量
- Stirring_Speed (rpm)：攪拌速度
- Product_Conc (mol/L)：產物濃度

```python
# 設定隨機種子
np.random.seed(SEED)

# 生成正常操作數據 (95%)
n_normal = 1900
n_anomaly = 100  # 5% 異常

# 正常操作數據（穩定操作條件，各變數彼此獨立）
normal_data = {
    'Temperature': np.random.normal(350, 8, n_normal),
    'Pressure': np.random.normal(8.0, 0.5, n_normal),
    'Feed_Flow': np.random.normal(35, 3, n_normal),
    'Cooling_Flow': np.random.normal(20, 2, n_normal),
    'Stirring_Speed': np.random.normal(300, 20, n_normal),
    'Product_Conc': np.random.normal(0.75, 0.08, n_normal),
}

# 5 種異常類型各 20 筆，每種異常只讓 2~4 個變數偏離正常值（軟異常）
# 異常類型 1: 溫度失控（溫度、壓力、冷卻流量上升，產物濃度略降）
temp_runaway = {
    'Temperature': np.random.normal(375, 8, 20),
    'Pressure': np.random.normal(8.8, 0.6, 20),
    'Feed_Flow': np.random.normal(35, 3, 20),
    'Cooling_Flow': np.random.normal(23, 2, 20),
    'Stirring_Speed': np.random.normal(300, 20, 20),
    'Product_Conc': np.random.normal(0.70, 0.08, 20),
}

# 異常類型 2: 壓力異常（壓力與進料流量上升）
pressure_anomaly = {
    'Temperature': np.random.normal(350, 8, 20),
    'Pressure': np.random.normal(9.2, 0.7, 20),
    'Feed_Flow': np.random.normal(38, 3, 20),
    'Cooling_Flow': np.random.normal(20, 2, 20),
    'Stirring_Speed': np.random.normal(300, 20, 20),
    'Product_Conc': np.random.normal(0.75, 0.08, 20),
}

# 異常類型 3: 冷卻流量異常（冷卻流量下降，溫度略升）
cooling_anomaly = {
    'Temperature': np.random.normal(362, 8, 20),
    'Pressure': np.random.normal(8.3, 0.5, 20),
    'Feed_Flow': np.random.normal(35, 3, 20),
    'Cooling_Flow': np.random.normal(15, 2, 20),
    'Stirring_Speed': np.random.normal(300, 20, 20),
    'Product_Conc': np.random.normal(0.72, 0.08, 20),
}

# 異常類型 4: 攪拌故障（攪拌速度與產物濃度下降）
stirring_fault = {
    'Temperature': np.random.normal(355, 8, 20),
    'Pressure': np.random.normal(8.0, 0.5, 20),
    'Feed_Flow': np.random.normal(35, 3, 20),
    'Cooling_Flow': np.random.normal(20, 2, 20),
    'Stirring_Speed': np.random.normal(220, 25, 20),
    'Product_Conc': np.random.normal(0.70, 0.08, 20),
}

# 異常類型 5: 產物濃度異常（產物濃度與進料流量下降）
product_anomaly = {
    'Temperature': np.random.normal(350, 8, 20),
    'Pressure': np.random.normal(8.0, 0.5, 20),
    'Feed_Flow': np.random.normal(32, 3, 20),
    'Cooling_Flow': np.random.normal(20, 2, 20),
    'Stirring_Speed': np.random.normal(300, 20, 20),
    'Product_Conc': np.random.normal(0.60, 0.08, 20),
}

# 合併所有數據
all_data = []
labels = []

# 添加正常數據
for i in range(n_normal):
    sample = [
        normal_data['Temperature'][i],
        normal_data['Pressure'][i],
        normal_data['Feed_Flow'][i],
        normal_data['Cooling_Flow'][i],
        normal_data['Stirring_Speed'][i],
        normal_data['Product_Conc'][i]
    ]
    all_data.append(sample)
    labels.append(0)  # 0 代表正常

# 添加異常數據
for anomaly_dict in [temp_runaway, pressure_anomaly, cooling_anomaly, 
                     stirring_fault, product_anomaly]:
    for i in range(20):
        sample = [
            anomaly_dict['Temperature'][i],
            anomaly_dict['Pressure'][i],
            anomaly_dict['Feed_Flow'][i],
            anomaly_dict['Cooling_Flow'][i],
            anomaly_dict['Stirring_Speed'][i],
            anomaly_dict['Product_Conc'][i]
        ]
        all_data.append(sample)
        labels.append(1)  # 1 代表異常

# 轉換為 NumPy 陣列
X = np.array(all_data)
y = np.array(labels)

# 建立 DataFrame
feature_names = ['Temperature', 'Pressure', 'Feed_Flow', 
                'Cooling_Flow', 'Stirring_Speed', 'Product_Conc']
df = pd.DataFrame(X, columns=feature_names)
df['Anomaly'] = y

# 顯示數據資訊
print("=== 數據集資訊 ===")
print(f"總樣本數: {len(df)}")
print(f"正常樣本數: {np.sum(y == 0)} ({np.sum(y == 0)/len(df)*100:.1f}%)")
print(f"異常樣本數: {np.sum(y == 1)} ({np.sum(y == 1)/len(df)*100:.1f}%)")
print(f"\n特徵數量: {X.shape[1]}")
print(f"特徵名稱: {feature_names}")

# 顯示數據統計比較
print("\n=== 正常 vs 異常數據統計比較 ===")
comparison = pd.DataFrame({
    'Normal_Mean': df[df['Anomaly'] == 0][feature_names].mean(),
    'Normal_Std': df[df['Anomaly'] == 0][feature_names].std(),
    'Anomaly_Mean': df[df['Anomaly'] == 1][feature_names].mean(),
    'Anomaly_Std': df[df['Anomaly'] == 1][feature_names].std(),
})
comparison['Difference'] = abs(comparison['Normal_Mean'] - comparison['Anomaly_Mean'])
print(comparison)

# 顯示前幾筆數據
print("\n=== 數據預覽 ===")
print(df.head(10))
```

## 2. 數據前處理

```python
# 分割訓練集與測試集
# 注意：孤立森林是非監督學習，訓練時不使用標籤
# 訓練集仍含約 5% 未標註的異常（污染訓練），標籤僅用於測試集評估
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=SEED, stratify=y
)

print("=== 數據分割 ===")
print(f"訓練集大小: {X_train.shape[0]} 樣本")
print(f"  正常: {np.sum(y_train == 0)} ({np.sum(y_train == 0)/len(y_train)*100:.1f}%)")
print(f"  異常: {np.sum(y_train == 1)} ({np.sum(y_train == 1)/len(y_train)*100:.1f}%)")
print(f"\n測試集大小: {X_test.shape[0]} 樣本")
print(f"  正常: {np.sum(y_test == 0)} ({np.sum(y_test == 0)/len(y_test)*100:.1f}%)")
print(f"  異常: {np.sum(y_test == 1)} ({np.sum(y_test == 1)/len(y_test)*100:.1f}%)")

# 孤立森林對特徵尺度不敏感，模型直接使用原始數據
# 標準化僅用於 PCA 視覺化（各變數單位與量級差異極大），scaler 只用訓練集擬合
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("\n=== 特徵標準化（僅供 PCA 視覺化使用） ===")
print("孤立森林模型使用原始數據訓練與預測（5.4 節將驗證標準化不影響結果）")
```

## 3. 建立與訓練孤立森林模型

### 3.1 建立孤立森林模型

**關鍵超參數設定**：
- `n_estimators=100`：使用 100 棵孤立樹
- `contamination=0.05`：預期 5% 的數據為異常（與真實異常率相同），用來決定決策閾值 `offset_`，不影響異常分數的排序
- `max_samples='auto'`：自動設為 min(256, n_samples)
- `random_state=42`：設定隨機種子以確保可重現性
- `n_jobs=-1`：使用所有 CPU 核心加速訓練

**注意**：本案例的訓練集包含約 5% 未標註的異常（污染訓練），模型完全不使用標籤；`contamination` 是閾值設定，不是模型從標籤學到的資訊。


```python
# 建立孤立森林模型
iso_forest = IsolationForest(
    n_estimators=100,           # 使用 100 棵樹
    max_samples='auto',         # 每棵樹使用 256 個樣本
    contamination=0.05,         # 預期 5% 的數據為異常
    random_state=SEED,          # 設定隨機種子
    n_jobs=-1,                  # 使用所有 CPU 核心
    verbose=0                   # 不顯示訓練過程
)

print("=== 孤立森林模型參數 ===")
print(f"樹的數量 (n_estimators): {iso_forest.n_estimators}")
print(f"子樣本大小 (max_samples): {iso_forest.max_samples}")
print(f"異常比例 (contamination): {iso_forest.contamination}")
print(f"隨機種子 (random_state): {iso_forest.random_state}")
print(f"並行運算 (n_jobs): {iso_forest.n_jobs}")
```

### 3.2 訓練模型

```python
import time

# 訓練模型
print("=== 開始訓練孤立森林模型 ===")
start_time = time.time()

iso_forest.fit(X_train)

training_time = time.time() - start_time
print(f"✓ 訓練完成！訓練時間: {training_time:.4f} 秒")

# 顯示模型屬性
print(f"\n=== 訓練後模型屬性 ===")
print(f"決策閾值偏移量 (offset_): {iso_forest.offset_:.4f}")
print(f"訓練特徵數量 (n_features_in_): {iso_forest.n_features_in_}")
print(f"孤立樹數量: {len(iso_forest.estimators_)}")
```

## 4. 異常檢測與評估

### 4.1 預測異常標籤與計算異常分數

```python
# 預測測試集
y_pred = iso_forest.predict(X_test)  # 1: 正常, -1: 異常
y_pred_binary = (y_pred == -1).astype(int)  # 轉換為 0: 正常, 1: 異常

# 計算異常分數
anomaly_scores = iso_forest.score_samples(X_test)  # 分數越低越異常
decision_scores = iso_forest.decision_function(X_test)  # 負值表示異常

print("=== 預測結果 ===")
print(f"測試集樣本數: {len(y_test)}")
print(f"預測為正常: {np.sum(y_pred == 1)} 樣本")
print(f"預測為異常: {np.sum(y_pred == -1)} 樣本")
print(f"異常比例: {np.sum(y_pred == -1)/len(y_pred)*100:.2f}%")

print(f"\n=== 異常分數統計 ===")
print(f"異常分數範圍: [{anomaly_scores.min():.4f}, {anomaly_scores.max():.4f}]")
print(f"異常分數平均: {anomaly_scores.mean():.4f}")
print(f"異常分數標準差: {anomaly_scores.std():.4f}")
print(f"決策閾值: {iso_forest.offset_:.4f}")
```

### 4.2 評估模型性能：混淆矩陣與分類報告

```python
# 計算混淆矩陣
cm = confusion_matrix(y_test, y_pred_binary)

print("=== 混淆矩陣 ===")
print(cm)
print(f"\nTrue Negatives (TN): {cm[0, 0]} - 正確識別為正常")
print(f"False Positives (FP): {cm[0, 1]} - 誤報為異常")
print(f"False Negatives (FN): {cm[1, 0]} - 漏報異常")
print(f"True Positives (TP): {cm[1, 1]} - 正確識別為異常")

# 視覺化混淆矩陣
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
            xticklabels=['Normal', 'Anomaly'],
            yticklabels=['Normal', 'Anomaly'],
            cbar_kws={'label': 'Count'})
plt.ylabel('True Label')
plt.xlabel('Predicted Label')
plt.title('Confusion Matrix - Isolation Forest', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig(FIG_DIR / 'confusion_matrix.png', dpi=300, bbox_inches='tight')
plt.show()

# 分類報告
print("\n=== 分類報告 ===")
print(classification_report(y_test, y_pred_binary, 
                          target_names=['Normal', 'Anomaly'],
                          digits=4))

# 計算關鍵指標
precision = precision_score(y_test, y_pred_binary)
recall = recall_score(y_test, y_pred_binary)
f1 = f1_score(y_test, y_pred_binary)

print("=== 關鍵指標摘要 ===")
print(f"Precision (精確率): {precision:.4f} - 告警中有多少是真正的異常")
print(f"Recall (召回率): {recall:.4f} - 實際異常中有多少被檢測到")
print(f"F1-Score: {f1:.4f} - 精確率與召回率的調和平均")

print("\n✓ 混淆矩陣已儲存至:", FIG_DIR / 'confusion_matrix.png')
```

### 4.3 ROC 曲線與 AUC

```python
# 計算 ROC 曲線（使用異常分數的負值，使異常點分數更高）
anomaly_scores_positive = -anomaly_scores
fpr, tpr, thresholds = roc_curve(y_test, anomaly_scores_positive)
roc_auc = auc(fpr, tpr)

# 繪製 ROC 曲線
plt.figure(figsize=(10, 8))
plt.plot(fpr, tpr, color='darkorange', lw=2.5, 
         label=f'Isolation Forest (AUC = {roc_auc:.3f})')
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--', 
         label='Random Classifier (AUC = 0.500)')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate (FPR)', fontsize=12)
plt.ylabel('True Positive Rate (TPR)', fontsize=12)
plt.title('ROC Curve - Isolation Forest Anomaly Detection', 
          fontsize=14, fontweight='bold')
plt.legend(loc="lower right", fontsize=11)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(FIG_DIR / 'roc_curve.png', dpi=300, bbox_inches='tight')
plt.show()

print(f"=== ROC-AUC 評估 ===")
print(f"AUC-ROC: {roc_auc:.4f}")
if roc_auc > 0.9:
    print("評價：優秀 (Excellent)")
elif roc_auc > 0.8:
    print("評價：良好 (Good)")
elif roc_auc > 0.7:
    print("評價：尚可 (Fair)")
else:
    print("評價：需改進 (Needs Improvement)")

print("\n✓ ROC 曲線已儲存至:", FIG_DIR / 'roc_curve.png')
```

### 5.3 contamination 敏感度分析

`contamination` 只決定決策閾值 `offset_`（訓練分數的第 $100 \times \mathrm{contamination}$ 百分位數），不改變異常分數的排序，因此 AUC 不會隨之改變。下面重新訓練不同 `contamination` 的模型（相同 `random_state`，樹的結構相同），觀察精確率與召回率的權衡。

```python
contamination_list = [0.01, 0.03, 0.05, 0.07, 0.10, 0.15]
rows = []

for c in contamination_list:
    m = IsolationForest(n_estimators=100, contamination=c,
                        random_state=SEED, n_jobs=-1).fit(X_train)
    pred = (m.predict(X_test) == -1).astype(int)
    rows.append({
        'contamination': c,
        'offset_': m.offset_,
        'flagged': int(pred.sum()),
        'precision': precision_score(y_test, pred),
        'recall': recall_score(y_test, pred),
        'f1': f1_score(y_test, pred),
        'auc': roc_auc_score(y_test, -m.score_samples(X_test)),
    })

cont_df = pd.DataFrame(rows)
print("=== contamination 敏感度（測試集 600 筆，其中 30 筆真實異常） ===")
print(cont_df.round(4).to_string(index=False))

# 繪圖：Precision / Recall / F1 隨 contamination 變化
plt.figure(figsize=(9, 6))
plt.plot(cont_df['contamination'], cont_df['precision'], 'o-', label='Precision', lw=2)
plt.plot(cont_df['contamination'], cont_df['recall'], 's-', label='Recall', lw=2)
plt.plot(cont_df['contamination'], cont_df['f1'], '^-', label='F1-Score', lw=2)
plt.axvline(0.05, color='gray', linestyle='--', label='True anomaly rate (5%)')
plt.xlabel('contamination', fontsize=12)
plt.ylabel('Score', fontsize=12)
plt.title('Effect of contamination on Precision / Recall / F1', fontsize=14, fontweight='bold')
plt.legend(fontsize=11)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(FIG_DIR / 'contamination_sensitivity.png', dpi=300, bbox_inches='tight')
plt.show()

print("\n✓ contamination 敏感度圖已儲存至:", FIG_DIR / 'contamination_sensitivity.png')
```

### 5.4 穩定性與尺度不變性檢查

孤立森林是隨機演算法，AUC 會隨 `random_state` 波動。下面對不同 `n_estimators` 與 `max_samples` 各用 10 個隨機種子重複訓練，觀察 AUC 的平均值與標準差；同時檢查標準化是否改變結果（孤立森林對每個特徵獨立做隨機切分，單調的尺度變換不影響切分結果）。

```python
n_seeds = 10

def auc_over_seeds(**params):
    """以不同隨機種子重複訓練，回傳測試集 AUC 的平均值與標準差"""
    aucs = []
    for s in range(n_seeds):
        m = IsolationForest(random_state=s, **params).fit(X_train)
        aucs.append(roc_auc_score(y_test, -m.score_samples(X_test)))
    return np.mean(aucs), np.std(aucs)

# (1) n_estimators 的影響
n_est_list = [10, 50, 100, 200, 300, 500]
est_res = [auc_over_seeds(n_estimators=n) for n in n_est_list]

# (2) max_samples 的影響（1.0 代表使用全部 1400 筆訓練樣本）
ms_list = [64, 128, 256, 512, 1.0]
ms_res = [auc_over_seeds(max_samples=ms) for ms in ms_list]

print(f"=== n_estimators 與 AUC（{n_seeds} 個隨機種子） ===")
for n, (mu, sd) in zip(n_est_list, est_res):
    print(f"n_estimators = {n:>3}: AUC = {mu:.4f} ± {sd:.4f}")

print(f"\n=== max_samples 與 AUC（{n_seeds} 個隨機種子） ===")
for ms, (mu, sd) in zip(ms_list, ms_res):
    print(f"max_samples = {str(ms):>5}: AUC = {mu:.4f} ± {sd:.4f}")

# (3) 標準化是否影響結果（相同 random_state）
m_raw = IsolationForest(random_state=SEED, contamination=0.05).fit(X_train)
m_scaled = IsolationForest(random_state=SEED, contamination=0.05).fit(X_train_scaled)
agree = np.mean(m_raw.predict(X_test) == m_scaled.predict(X_test_scaled))
auc_raw = roc_auc_score(y_test, -m_raw.score_samples(X_test))
auc_scaled = roc_auc_score(y_test, -m_scaled.score_samples(X_test_scaled))
print("\n=== 標準化前後比較（random_state=42） ===")
print(f"原始數據 AUC: {auc_raw:.4f}")
print(f"標準化數據 AUC: {auc_scaled:.4f}")
print(f"預測標籤一致率: {agree:.2%}")

# 繪圖
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].errorbar(range(len(n_est_list)), [r[0] for r in est_res],
                 yerr=[r[1] for r in est_res], fmt='o-', capsize=5, lw=2)
axes[0].set_xticks(range(len(n_est_list)))
axes[0].set_xticklabels(n_est_list)
axes[0].set_xlabel('n_estimators', fontsize=12)
axes[0].set_ylabel('Test AUC (mean ± std)', fontsize=12)
axes[0].set_title('AUC vs n_estimators', fontsize=13, fontweight='bold')
axes[0].grid(alpha=0.3)

axes[1].errorbar(range(len(ms_list)), [r[0] for r in ms_res],
                 yerr=[r[1] for r in ms_res], fmt='s-', capsize=5, lw=2, color='darkorange')
axes[1].set_xticks(range(len(ms_list)))
axes[1].set_xticklabels([str(v) for v in ms_list])
axes[1].set_xlabel('max_samples', fontsize=12)
axes[1].set_ylabel('Test AUC (mean ± std)', fontsize=12)
axes[1].set_title('AUC vs max_samples', fontsize=13, fontweight='bold')
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.savefig(FIG_DIR / 'stability_check.png', dpi=300, bbox_inches='tight')
plt.show()

print("\n✓ 穩定性檢查圖已儲存至:", FIG_DIR / 'stability_check.png')
```

### 5.5 特徵重要性：Permutation Importance

孤立森林每次切分都是從所有特徵中均勻隨機挑選，因此「統計各特徵被用來切分的次數」會趨近均勻分布，無法反映特徵的重要性。較可靠的做法是 permutation importance：在測試集上隨機打亂單一特徵，觀察 AUC 下降多少（下降越多，表示模型越依賴該特徵）。

```python
rng = np.random.RandomState(SEED)
n_repeats = 20

# (1) Permutation importance：打亂單一特徵後 AUC 的下降量
base_auc = roc_auc_score(y_test, -iso_forest.score_samples(X_test))
perm_mean, perm_std = [], []
for j in range(X_test.shape[1]):
    drops = []
    for _ in range(n_repeats):
        X_perm = X_test.copy()
        X_perm[:, j] = rng.permutation(X_perm[:, j])
        drops.append(base_auc - roc_auc_score(y_test, -iso_forest.score_samples(X_perm)))
    perm_mean.append(np.mean(drops))
    perm_std.append(np.std(drops))

# (2) 對照：各特徵被用於切分的次數比例（內部節點）
split_counts = np.zeros(len(feature_names))
for tree, feats in zip(iso_forest.estimators_, iso_forest.estimators_features_):
    used = tree.tree_.feature[tree.tree_.feature >= 0]
    for f in used:
        split_counts[feats[f]] += 1
split_ratio = split_counts / split_counts.sum()

imp_df = pd.DataFrame({
    'Feature': feature_names,
    'AUC_drop_mean': perm_mean,
    'AUC_drop_std': perm_std,
    'Split_ratio': split_ratio,
}).sort_values('AUC_drop_mean', ascending=False)

print(f"=== 特徵重要性（基準 AUC = {base_auc:.4f}，重複 {n_repeats} 次） ===")
print(imp_df.round(4).to_string(index=False))

# 繪圖
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].barh(imp_df['Feature'], imp_df['AUC_drop_mean'], xerr=imp_df['AUC_drop_std'],
             color='steelblue', capsize=4)
axes[0].invert_yaxis()
axes[0].set_xlabel('AUC drop after permutation', fontsize=12)
axes[0].set_title('Permutation Importance', fontsize=13, fontweight='bold')
axes[0].grid(alpha=0.3)

axes[1].barh(imp_df['Feature'], imp_df['Split_ratio'], color='gray')
axes[1].axvline(1 / len(feature_names), color='red', linestyle='--', label='Uniform (1/6)')
axes[1].invert_yaxis()
axes[1].set_xlabel('Share of splits', fontsize=12)
axes[1].set_title('Split Frequency (not informative)', fontsize=13, fontweight='bold')
axes[1].legend()
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.savefig(FIG_DIR / 'feature_importance.png', dpi=300, bbox_inches='tight')
plt.show()

print("\n✓ 特徵重要性圖已儲存至:", FIG_DIR / 'feature_importance.png')
```

### 6.1 孤立森林核心概念

通過本 Notebook 的實作，我們學習了：

1. **孤立森林的核心理念**：
   - 異常點更容易被「孤立」，需要較少的隨機分割次數
   - 基於路徑長度而非距離或密度進行異常檢測
   - 適合高維大規模數據的異常檢測

2. **關鍵超參數設定**：
   - `n_estimators`：樹的數量（通常 100-300）
   - `contamination`：預期異常比例（需根據領域知識設定）
   - `max_samples`：每棵樹的子樣本大小（預設 'auto' = min(256, n_samples)）

3. **模型評估方法**：
   - 有標籤時：精確率、召回率、F1、AUC-ROC
   - 無標籤時：視覺化、異常分數分布、領域知識驗證

4. **化工應用場景**：
   - 反應器異常監控
   - 產品品質檢測
   - 設備健康監控
   - 製程安全預警

### 6.2 實務應用建議

**化工製程異常檢測的部署策略**：

1. **數據收集階段**：
   - 收集至少 1,000 筆正常操作數據
   - 確保數據涵蓋各種正常操作條件
   - 記錄所有已知的異常事件

2. **模型訓練階段**：
   - 使用正常操作數據訓練孤立森林
   - 根據歷史異常率設定 `contamination` 參數
   - 使用交叉驗證調整超參數

3. **離線驗證階段**：
   - 使用歷史異常事件驗證模型性能
   - 分析誤報與漏報案例
   - 與製程工程師討論結果的合理性

4. **在線部署階段**：
   - 建立即時數據流處理管道
   - 實現告警系統（設定告警抑制策略）

   - 定期重新訓練模型（每月或每季）
   - 整合操作員反饋，調整參數
   - 監控模型性能變化

**常見陷阱與解決方案**：

| 問題 | 原因 | 解決方案 |
|------|------|----------|
